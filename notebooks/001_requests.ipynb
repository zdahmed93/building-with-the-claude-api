# 01 · Your first message to Claude

Before you start, copy `.env.example` to `.env` and paste your API key into it.

## 1. Load your API key from `.env`

```typescript
import dotenv from "npm:dotenv@18.0.9";

dotenv.config({ path: "../.env" }); // the notebook runs from notebooks/, so .env is one level up

// Check the key is loaded without printing it
console.log(process.env.ANTHROPIC_API_KEY ? "API key found" : "API key missing — check your .env file");
```

## 2. Send a message

```typescript
import Anthropic from "npm:@anthropic-ai/sdk@1.3.0";

const client = new Anthropic(); // reads ANTHROPIC_API_KEY automatically

const response = await client.messages.create({
  model: "claude-haiku-5-5",
  max_tokens: 1024,
  messages: [{ role: "user", content: "AI in one sentence" }],
});

response
```

## 3. Print the reply

```typescript
let answer = response.content[0].text
answer
```

# Messages History

```typescript
const addMessage = (role: "user" | "assistant", messages: any[], text: string) => {
    messages.push({
        role,
        content: text
    })
}
```

```typescript
const chat = async (messages: any[]) => {
    const response = await client.messages.create({
        model: "claude-haiku-5-5",
        max_tokens: 1024,
        messages: [{ role: "user", content: "AI in one sentence" }],
    });
    return response.content[0].text
}
```

```typescript
const messages = []
addMessage("user", messages, "AI in one sentence")
messages
```

```typescript
addMessage("assistant", messages, answer)
messages
```

```typescript
addMessage("user", messages, "Give me another sentence about the same subject")
messages
```

```typescript
answer = await chat(messages)
answer
```
